# Ex4.1

$H_0$: Null hypotheis

$H_1$: Alt hypothesis

False Positive $\alpha$: probability of making a false positive error. Rejecting $H_0$ when it is true

True Negative ($1 - \alpha$): Taking $H_0$

True Positive Power ($1 - \beta$): probability of correctly rejecting $H_0$, correctly taking $H_1$ 

False negative ($\beta$): taking $H_0$, incorrectly rejecting $H_1$

TP: Taking $H_1$

FP: Taking $H_1$, rejecting $H_0$ wrongly

TN: Taking $H_0$

FN: Taking $H_0$, rejecting $H_1$ wrongly


Prob of rejecting $H_0$ When it is true ($H_1$ incrorrectly taken): 0.05
Prob of rejecting $H_0$ when it is false ($H_1$ correctly taken): 0.95
