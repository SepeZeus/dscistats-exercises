### FDR Benjamini-Hochberg
1. order from smallest to largest (1-4, rank)

2. current p-value * (total p vals / pvalue rank)

3. choose smaller of 2 options (previous adjusted or new adjusted)

$$p_{\text{FDR}(i)} = p_{(i)} \cdot \frac{m}{i}$$

$$\hat{p}_j = \frac{1}{n} \sum_{i=1}^{n} I(X_i = j)$$Breakdown of the Notation:

$\hat{p}_j$: The estimated probability that an observation falls into category $j$.

$n$: The total number of observations (sample size).

$\sum_{i=1}^{n}$: A summation across all sample data points from $i = 1$ to $n$.

$I(X_i = j)$: An indicator function, which equals $1$ if the observation $X_i$ is equal to category $j$, and $0$ otherwise.
