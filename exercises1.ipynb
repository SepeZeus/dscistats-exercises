```python
#2.1
from scipy import stats
import math
import numpy as np
import matplotlib.pyplot as plt

X = stats.norm(loc=0, scale=1)
x = X.rvs(size=10000)
y = np.exp(x)
sample_mean_y= np.mean(y)
sample_var = np.var(y, ddof=1)

sample_mean_y, sample_var
```

```python
lognorm = stats.lognorm(s=1, scale=np.exp(0))
logn_x = lognorm.rvs(10000)
logn_mean = np.mean(logn_x)
logn_var = np.var(logn_x, ddof=1)
logn_mean, logn_var
```

```python
expon = stats.expon(scale=1)
expon_x = expon.rvs(10000)
expon_mean = np.mean(expon_x)
expon_var = np.var(expon_x, ddof=1)
expon_mean, expon_var
```

```python
gamma = stats.gamma(a=1, scale=1)
gamma_x = gamma.rvs(10000)
gamma_mean = np.mean(gamma_x)
gamma_var = np.var(gamma_x, ddof=1)
gamma_mean, gamma_var
```

```python
# Histogram (density=True normalizes it so the area sums to 1)
grid = np.linspace(0, 10, 1000)
plt.plot(grid, lognorm.pdf(grid), label=f'PDF: {"LogNorm"}', linewidth=2)
plt.plot(grid, expon.pdf(grid), label=f'PDF: {"Exponential"}', linewidth=2)
plt.plot(grid, gamma.pdf(grid), label=f'PDF: {"Gamma"}', linewidth=2)
plt.legend()
plt.xlabel('y')
plt.ylabel('Density')
plt.grid(True, alpha=0.3)
plt.title('Comparison of Sample Y Histogram against Theoretical PDFs')
plt.hist(y, bins=60, density=True, range=(0, 12), alpha=0.4, label='Sample Y histogram', color='gray')

#The distribution looks like a lognormal one based on the mean and variance values, and the histogram
#I only learned at the end that I could've done lognorm.mean(),expon.mean(), and gamma.mean()
#I used AI to help me with the syntax for scipy and the plot, and to help me understand the results
```
