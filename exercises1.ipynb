```python
#2.1
from scipy import stats
import math
import numpy as np
import matplotlib.pyplot as plt

X = stats.norm(loc=0, scale=1)
x = X.rvs(size=10000)
y = np.exp(x)
sample_mean_y= np.mean(y)
sample_var = np.var(y, ddof=1)

sample_mean_y, sample_var
```

```python
lognorm = stats.lognorm(s=1, scale=np.exp(0))
logn_x = lognorm.rvs(10000)
logn_mean = np.mean(logn_x)
logn_var = np.var(logn_x, ddof=1)
logn_mean, logn_var
```

```python
expon = stats.expon(scale=1)
expon_x = expon.rvs(10000)
expon_mean = np.mean(expon_x)
expon_var = np.var(expon_x, ddof=1)
expon_mean, expon_var
```

```python
gamma = stats.gamma(a=1, scale=1)
gamma_x = gamma.rvs(10000)
gamma_mean = np.mean(gamma_x)
gamma_var = np.var(gamma_x, ddof=1)
gamma_mean, gamma_var
```

```python
#3
import numpy as np
from scipy import stats
X = stats.uniform(loc=-1, scale=2)
x = X.rvs(10000)
y= x**2
Ex = np.mean(x)
Ey= np.mean(y)
Exy = np.mean(x*y)
CovXY = Exy - (Ex*Ey)

print(f"E[X]:   {Ex:.4f}")
print(f"E[Y]:   {Ey:.4f}")
print(f"E[XY]:  {Exy:.4f}")
print(f"Cov(X,Y): {CovXY:.4f}")

#E[XY] appears to be very close to Cov(X, Y), and they're both very close to 0.
#The mean of XY and the Covariance have near identical values
#E[Y] despite being only the square of E[X] is significantly larger than E[X}
#Despite the lack of correlation, X and Y are not independent, as Y is X^2,
#so directly dependent on X

```

```python
#4.
import numpy as np
from scipy import stats
X = stats.multivariate_normal([0, 1], [[1.0, 0.5], [0.5, 2.0]])
x = X.rvs(1000)
mean = np.mean(x[:, 0] + x[:, 1])
var = np.var(x[:, 0] + x[:, 1])
marg_mean = np.mean(x[:, 0]) + np.mean(x[:, 1])
marg_var = np.var(x[:, 0]) + np.var(x[:, 1])
mean, var, marg_mean, marg_var

#The mean of X_1+X_2 is equal to the sum of the marginal means of X_1 and X_2.
#The variances look mostly similar, however, the variance of X_1+X_2 is ~1.0 more
#than the sum of the marginal variances. According to AI this happens due to the math
#ending up being 2x the covariance, i.e. Var (X_1+X_2) - (Var(X_1)+Var(X_2)) = 2Cov(X_1,X_2)
#showing that 2*0.5 = 1.0.
#Due to positive covariances, X_1 and X_2 are dependent on one another.
#This code I wrote myself with the help of the scipy documentation. I asked AI to verify it.
```
